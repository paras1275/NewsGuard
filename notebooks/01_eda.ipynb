# NewsGuard — 01. Exploratory Data Analysis (EDA)
This notebook performs exploratory analysis on the NewsGuard fake news dataset, inspecting missing values, class balance, word count distributions, and subject categorizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/raw/news_dataset.csv')
print(f"Total Samples: {len(df)}")
df.head()
```

```python
# Class distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='label', palette='Set2')
plt.title('Target Label Distribution (0=FAKE, 1=REAL)')
plt.show()
```

```python
# Text Length Distribution
df['word_count'] = df['text'].apply(lambda x: len(str(x).split()))
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='word_count', hue='label', kde=True, bins=30)
plt.title('Word Count Distribution by Class')
plt.show()
```
